# Cleanup Old Styled Components Files

## 1. Prerequisites & setup
- Install required tools if needed:
  - `npm install -D depcheck jscodeshift ts-morph eslint-plugin-unused-imports`
  - Install `rg` (ripgrep) on your system.

## 2. Create backup git branch
- `git checkout -b chore/remove-old-styled-components`

## 3. Locate styled-components files
- Search candidates:
  - `rg "\.styled\.|styled\(" --glob "*.{js,jsx,ts,tsx}" -n`
  - `find . -name "*.styled.*" -print`

## 4. Static analysis: detect unused styled exports
- Use TS/JS AST to list styled exports and cross-check imports.
- Pseudocode: parse files -> collect exports -> search imports -> report unused.

## 5. Run depcheck and other unused-file tools
- `npx depcheck --json`
- Use eslint unused-imports and `rg` to verify unused files/references.

## 6. Codemod to update/remove imports and indexes
- Use `jscodeshift` or `ts-morph` to remove imports referencing styled files.
- Example: `npx jscodeshift -t transform/remove-styled-imports.js src`

## 7. Safe removal script (move to backup, git rm)
- Move candidates to backup first:
  - `mkdir -p .trash/removed-styled && git mv path/to/file .trash/removed-styled/`
- Run tests before final removal.

## 8. Run tests, lint, and type checks
- `npm test && npm run lint && npm run typecheck`

## 9. Verify build and CI locally
- `npm run build`
- Optionally run CI workflow locally with `act` or smoke tests.

## 10. Commit, push, open PR
- `git add -A`
- `git commit -m "chore: remove old styled-components files"`
- `git push -u origin HEAD`

## 11. Final cleanup (empty dirs, lockfile)
- `find . -type d -empty -delete`
- `npm prune`
- `npm install --package-lock-only`
- `npx prettier --write .`
